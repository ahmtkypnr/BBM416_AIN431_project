```python
import os
import shutil

# Kaynak klasörler
images_src = "srdatasetfixed/output_bilinear/output_bilinear/test"
labels_src = "dataset/labels/test"

# Hedef klasörler
images_dst = "srdatasetfixed_yolo_readyBilinear/images/test"
labels_dst = "srdatasetfixed_yolo_readyBilinear/labels/test"

# Hedef klasörleri oluştur
os.makedirs(images_dst, exist_ok=True)
os.makedirs(labels_dst, exist_ok=True)

# Görselleri kopyala
for fname in os.listdir(images_src):
    if fname.endswith(".jpg") or fname.endswith(".png"):
        shutil.copy(os.path.join(images_src, fname), os.path.join(images_dst, fname))

# Etiketleri kopyala
for fname in os.listdir(labels_src):
    if fname.endswith(".txt"):
        shutil.copy(os.path.join(labels_src, fname), os.path.join(labels_dst, fname))

print("✅ Dataset YOLO formatına dönüştürüldü.")
```
